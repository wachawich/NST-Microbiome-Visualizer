## ตอนที่ 1: การศึกษาด้วย Stacked Bar Plot

```python
# สร้าง Stacked Bar Plot ระดับ Phylum และตัด 1% Cutoff
plt_phylum = visualizer.stacked_bar(taxonomic_rank='phylum', percent_cutoff=1.0)
plt_phylum.show()
```

```python
# สร้าง Stacked Bar Plot ระดับ Genus และตัด 5% Cutoff
plt_genus = visualizer.stacked_bar(taxonomic_rank='genus', percent_cutoff=5.0)
plt_genus.show()
```
