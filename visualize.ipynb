```python
%load_ext autoreload
%autoreload 2
import matplotlib.pyplot as plt
from main import MicrobiomeVisualizer

visualizer = MicrobiomeVisualizer('feature_table.csv')
```

## ตอนที่ 1: การศึกษาด้วย Stacked Bar Plot

```python
# สร้าง Stacked Bar Plot ระดับ Phylum และตัด 1% Cutoff
plt_phylum = visualizer.stacked_bar(taxonomic_rank='phylum', percent_cutoff=1.0)
plt_phylum.show()
```

```python
# สร้าง Stacked Bar Plot ระดับ Genus และตัด 5% Cutoff
plt_genus = visualizer.stacked_bar(taxonomic_rank='genus', percent_cutoff=5.0)
plt_genus.show()
```

### Export to File

```python
visualizer.export_stacked_bar(taxonomic_rank='genus', file_format='png', percent_cutoff=5.0)
```

## ตอนที่ 2: การศึกษาด้วย Heatmap

```python
# Heatmap ระดับ Family + Z-score scaling + Clustering (default)
plt_hm = visualizer.heatmap(taxonomic_rank='family', percent_cutoff=2.0,
                            scaling='z-score', clustering=True)
plt_hm.show()
```

```python
# Heatmap ระดับ Phylum + log scaling + ไม่ใช้ clustering
plt_hm2 = visualizer.heatmap(taxonomic_rank='phylum', percent_cutoff=1.0,
                             scaling='log', clustering=False)
plt_hm2.show()
```

### Export Heatmap to File

```python
visualizer.export_heatmap(taxonomic_rank='family', percent_cutoff=2.0,
                          scaling='z-score', clustering=True,
                          file_format='png', filename='heatmap_family')
```
